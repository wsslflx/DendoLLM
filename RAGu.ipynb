```python
EMAIL = os.getenv("EMAIL", 'trifonova.kate.s@gmail.com')
BASE_URL = "https://api.openalex.org/works"
def works_with_oa(query, first_n=1):
    params = {
        'filter': f'abstract.search:"{query}",best_open_version:published',
        'per-page': first_n,
        # request publication year for downstream citations
        'select':'id,title,publication_year,best_oa_location',
        'mailto': EMAIL
    }
    papers = requests.get(BASE_URL, params=params).json()['results']
    return papers
def get_pdf(paper: dict, location: str) -> pathlib.Path | None:
    url = paper.get('best_oa_location', {}).get('pdf_url')
    paper_id = paper['id'].split('/')[-1]
    pdf_dir = pathlib.Path(location)
    pdf_dir.mkdir(parents=True, exist_ok=True)
    pdf_path = pdf_dir / f"{paper_id}.pdf"
    if not url:
        print(f'No PDF URL for paper {paper_id}')
        return None
    # Skip existing files
    if pdf_path.exists():
        return pdf_path
    # Attempt download with error handling
    with requests.Session() as session:
        session.headers.update({'From': EMAIL})
        try:
            resp = session.get(url, stream=True)
            resp.raise_for_status()
        except requests.HTTPError as e:
            print(f"HTTPError downloading PDF for paper {paper_id}: {e}")
            return None
        except Exception as e:
            print(f"Error downloading PDF for paper {paper_id}: {e}")
            return None
        # Write content in chunks
        with open(pdf_path, 'wb') as f:
            for chunk in resp.iter_content(chunk_size=8192):
                if chunk:
                    f.write(chunk)
    return pdf_path
```

```python
def has_text(pdf_path: Union[str, pathlib.Path], max_pages: int = 3, min_words: int = 50) -> bool:
    reader = PdfReader(pdf_path)
    word_count = 0

    for i, page in enumerate(reader.pages):
        if i >= max_pages:
            break
        text = page.extract_text()
        if text:
            words = text.strip().split()
            word_count += len(words)
            if word_count >= min_words:
                return True

    return False

def load_and_split_docs(pdf_path: pathlib.Path) -> list:

    if has_text(pdf_path):
        loader = PyPDFLoader(str(pdf_path))
    else:
        loader = UnstructuredPDFLoader(str(pdf_path), strategy="ocr_only")

    docs = loader.load()
    splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
    splits = splitter.split_documents(docs)

    for i, doc in enumerate(splits):
        doc.metadata.update({
            "chunk_index": i,
            "source_path": str(pdf_path)
        })

    return splits
```

```python
def fetch_and_prepare(query: str, location: str = './pdfs', first_new: int = 3) -> list:

    pdf_dir = pathlib.Path(location)
    pdf_dir.mkdir(parents=True, exist_ok=True)

    # Over-fetch candidates (3x) to allow skipping existing files
    candidates = works_with_oa(query, first_new * 3)
    new_docs = []

    for paper in candidates:
        paper_id = paper['id'].split('/')[-1]
        pdf_path = pdf_dir / f"{paper_id}.pdf"
        # Skip if already downloaded
        if pdf_path.exists():
            continue
        # Download and prepare
        downloaded = get_pdf(paper, location)
        if downloaded:
            new_docs.extend(load_and_split_docs(downloaded))
        # Stop once we've got enough new ones
        if len(new_docs) >= first_new:
            break
    return new_docs
```

```python
class RAG:

    def __init__(self):
        self.vectorstore = Chroma(collection_name="bunch_of_docs", embedding_function=OpenAIEmbeddings())
        self.threshold = 0.5
        self.ingested_per_species = defaultdict(set)

    @staticmethod
    def has_text(pdf_path: str, max_pages: int = 3, min_words: int = 10) -> bool:
        reader = PdfReader(pdf_path)
        word_count = 0
        for i, page in enumerate(reader.pages):
            if i >= max_pages:
                break
            text = page.extract_text()
            if text:
                words = text.strip().split()
                word_count += len(words)
                if word_count >= min_words:
                    return True
        return False

    @staticmethod
    def fetch_and_prepare(query: str, location: str = './pdfs', first_new: int = 10) -> list:

        pdf_dir = pathlib.Path(location)
        pdf_dir.mkdir(parents=True, exist_ok=True)

        # Over-fetch candidates (3x) to allow skipping existing files
        candidates = works_with_oa(query, first_new * 3)
        downloaded_entries = []

        for paper in candidates:
            # keep both PDF path and OpenAlex metadata together
            paper_id = paper['id'].split('/')[-1]
            pdf_path = pdf_dir / f"{paper_id}.pdf"

            # Skip if already downloaded
            if pdf_path.exists():
                continue
            try:
                downloaded = get_pdf(paper, location)
                if downloaded:
                    downloaded_entries.append({
                        "pdf_path": downloaded,
                        "paper": paper
                    })

            except Exception as e:
                print(f"Failed to download PDF for paper {paper['id']}: {e}")
                continue

            if len(downloaded_entries) >= first_new:
                break

        return downloaded_entries

    def load_ocr(self, pdf_path: str, puppy: str = None, paper_meta: Optional[dict] = None):
        if self.has_text(pdf_path):
            print("Text detected in PDF. Using fast strategy.")
            loader = PyPDFLoader(pdf_path)
        else:
            print("No text detected. Using OCR.")
            loader = UnstructuredPDFLoader(pdf_path, strategy="ocr_only")

        docs = loader.load()
        splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
        splits = splitter.split_documents(docs)

        for i, doc in enumerate(splits):
            # merge loader metadata with paper details before storing
            metadata = dict(doc.metadata) if doc.metadata else {}
            # annotate each chunk with local bookkeeping info
            metadata.update({
                "chunk_index": i,
                "source_path": str(pdf_path),
                "specie": puppy
            })
            if paper_meta:
                # attach bibliographic metadata for later citation
                metadata.update({
                    "openalex_id": paper_meta.get("id"),
                    "title": paper_meta.get("title"),
                    "publication_year": paper_meta.get("publication_year"),
                    "pdf_url": paper_meta.get("best_oa_location", {}).get("pdf_url")
                })
            doc.metadata = metadata

        self.vectorstore.add_documents(splits)
        return splits

    def chain(self, question):
        if not self.vectorstore:
            return "Vector store not initialized."

        def format_docs(docs):
            return "\n".join(doc.page_content for doc in docs)

        results_with_scores = self.vectorstore.similarity_search_with_score(question, k=20)
        filtered_docs = [doc for doc, score in results_with_scores if score <= self.threshold]

        if not filtered_docs:
            return "booooooo, no docs found within the given threshold"

        used_per_species = defaultdict(set)
        for doc in filtered_docs:
            specie = (doc.metadata or {}).get("specie", "").strip()
            source = (doc.metadata or {}).get("source_path")
            if specie and source:
                used_per_species[specie].add(source)

        prompt = PromptTemplate(
            input_variables=["context", "question"],
            template=pathlib.Path('prompt.txt').read_text(encoding='utf8')
        )

        llm = ChatOpenAI(model_name="gpt-3.5-turbo", temperature=0.2)
        def inspect_prompt(inputs):
            formatted_prompt = prompt.format(context=format_docs(filtered_docs), question=question)
            print(formatted_prompt)
            return inputs

        rag_chain = (
            {"context": RunnableLambda(lambda _: format_docs(filtered_docs)),
             "question": RunnablePassthrough()}
            | RunnableLambda(inspect_prompt)
            | prompt
            | llm
            | StrOutputParser()
        )

        answer = rag_chain.invoke(question)
        summaries = []
        all_species = set(self.ingested_per_species.keys()) | set(used_per_species.keys())
        for specie in sorted(all_species):
            ingested = len(self.ingested_per_species.get(specie, set()))
            used = len(used_per_species.get(specie, set()))
            summaries.append(f"{specie}: {ingested} PDFs ingested / {used} used")
        counts_line = "Counts: " + "; ".join(summaries) if summaries else "Counts: none"

        return f"{answer}{counts_line}"

    def query(self, question: str):
        self.ingested_per_species.clear()
        for puppy in question.split(","):
            papers = self.fetch_and_prepare(query=puppy)

            for paper_entry in papers:
                pdf_path = paper_entry["pdf_path"]
                paper_meta = paper_entry.get("paper")
                print(f'Processing PDF: {pdf_path}')
                # pass metadata so chunks retain citation fields
                self.load_ocr(pdf_path=str(pdf_path), puppy=puppy.strip(), paper_meta=paper_meta)
                self.ingested_per_species[puppy.strip()].add(str(pdf_path))

        print(self.chain(question))


```
